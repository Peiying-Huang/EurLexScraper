The DocumentInfoScraper extracts the following structure.\

The normal structure of the website:
1. Dates : [dict1, dict2, dict3,...]
2. Miscellaneous information:[dict1, dict2, dict3,...]
3. Procedure:[dict1, dict2, dict3,...]
4. Relationship between documents:[dict1, dict2, dict3,...] (if the text is a hyperlink, the text and the link will be stored as a dictnary as the value of dictN)
5. Classification:[dict1, dict2, dict3,...] (if the text is a hyperlink, the text and the link will be stored as a dictnary as the value of dictN)\

Some of the websites only contains parts of the sections, e.g. ['Dates', 'Miscellaneous information', 'Procedure']. If the website doesn't exist, the code returns an empty dictionary {}.
