The DocumentInfoScraper extracts the following structure. If the website doesn't exist, the code returns an empty dictionary {}.\
The normal structure of the website:
1. Dates : [dict1, dict2, dict3,...]
2. Miscellaneous information:[dict1, dict2, dict3,...]
3. Procedure:[dict1, dict2, dict3,...]
4. Relationshi between documents:[dict1, dict2, dict3,...]
5. Classification:[dict1, dict2, dict3,...]
(if the text is a hyperlink, the text and the link will be stored as a dictnary as the value of dictN)
